```python
def getPositiveNumber(askForIt):
  wageFlag = False
  while not wageFlag:
    hourWage = float(input(askForIt))
    if hourWage >= 0:
      print("")
      wageFlag = True
    else:
      print("Try again within a normal range")
  return hourWage

def getNumberOfDays():
  workingFlag = False
  while not workingFlag:
    daysWorking = int(input("How many days are you working this week? "))
    if 1 <= daysWorking <= 7:
      print("")
      workingFlag = True
    else:
      print("Try again within a normal range")
  return daysWorking

def collectDailyHours(daysWorking):
  dailyHours = []
  for day in range(1, daysWorking + 1):
    hours = getPositiveNumber(f"Hours worked on day {day}: ")
    dailyHours.append(hours)
  return dailyHours

def calcTotalHours(dailyHours):
  total = 0
  for hours in dailyHours:
    total += hours
  return total

def calcPay(dailyHours, hourlyWage):
  totalPay = 0
  hoursNow = 0
  for hours in dailyHours:
    regularDailyHours = min(hours, 12)
    doubleOThours = max(0, hours - 12)
    if hoursNow >= 40:
      totalPay += regularDailyHours * hourlyWage * 1.5
    elif hoursNow + regularDailyHours > 40:
      hoursReg = 40 - hoursNow
      hoursOT = regularDailyHours - hoursReg
      totalPay += hoursReg * hourlyWage
      totalPay += hoursOT * hourlyWage * 1.5
    else:
      totalPay += regularDailyHours * hourlyWage
    totalPay += doubleOThours * hourlyWage * 2
    hoursNow += hours
  return totalPay

daysWorking = getNumberOfDays()
hourlyWage = getPositiveNumber("What is your hourly wage? $")

dailyHours = collectDailyHours(daysWorking)
totalHours = calcTotalHours(dailyHours)
weeklyPay = calcPay(dailyHours, hourlyWage)

print(f"{totalHours} Total hours worked ")
print(f"Weekly pay: {(weeklyPay):.2f}")

```
