### Problem 1: Personalized Greeting
Write a function called `miniMadLib` that takes three arguments `noun`, `verb`, and `adjective` and combines them to print out a one sentence madlib.

```python
#items are the parameters of the function miniMadLib
def miniMadLib(noun, verb, adjective):
  noun = input("what is your noun? ")
  verb = input("what is your verb? ")
  adjective = input("what is your adjective? ")
  sum = noun + verb + adjective
  return print(f"{sum}")


miniMadLib(2,2,2)
```

### Problem 2: Calculate the Volume of a Cylinder
Write a function called `calculateCylinderVolume` that takes the radius and height of a cylinder and returns its internal volume. Use the built in math constant for pi.

```python
from math import pi

def calculateCylinderVolume():
  radius = int(input("How much is the radius? "))
  height = int(input("What is the height of the cylinder? "))
  print(f"Calculated the volume of a cylinder with a radius of {radius} and a height of {height}")
  return print(f"The volume of a cylinder is {((pi*radius*radius)*height):.2f}")

calculateCylinderVolume()
```
