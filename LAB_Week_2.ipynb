```python
#Kilometer to Mile Conversion Tool
kilometer = float(input("Please enter the Kilometer(s) amount to  convert to miles: "))
mile = kilometer * 0.621371
print(f"{kilometer} kilometers is {mile} miles")
```

```python
#Celsius to Fahrenheit Conversion Tool
c = float(input("Pleast enter the C amount to covert to F temperature: "))
f = (c * (9/5)) + 32
print(f"{c} C is {f} F temperature")
```
